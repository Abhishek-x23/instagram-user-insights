```python
with open("finaldata.txt",encoding='utf-8') as f:
    data=f.read()
```

```python
    
    chunks=data.split("\n\n")
    chunks=[c for c in chunks if len(c)>3]
    
```

```python
def parse_allNumbers_toInt(sep_chunk,split_name):
    num= float(sep_chunk.split(split_name)[0].replace(",","").replace("K","").replace("M",""))
    if("K" in sep_chunk):
        num=int(num*1000)
    elif("M" in sep_chunk):
        num=int(num*1000000)
    else:
        num=int(num)
    return num
```

```python
def parse_chunks(chunk):
    chunk=chunk.strip()
    sep_chunk=chunk.split("\n")
    username=sep_chunk[0]
    posts=parse_allNumbers_toInt(sep_chunk[1]," post")
    
    followers=parse_allNumbers_toInt(sep_chunk[2]," follower")
    following=parse_allNumbers_toInt(sep_chunk[3]," following")
    name=sep_chunk[4]
    if(len(sep_chunk)>5):
        page_type=sep_chunk[5]
        bio="\n".join(sep_chunk[6:])
    else:
        page_type="Unknown"
        bio="Unknown"
    return{"username":username,"posts":posts,"followers":followers,"following":following,"name":name,"page_type":page_type,"bio":bio}
```

```python
chunkies=[]
for c in chunks:
    new_chunk= parse_chunks(c)
    chunkies.append(new_chunk)
print(chunkies)
```

```python
import json
s=json.dumps(chunkies,indent=4)
with open("finalOutput.json","w") as f:
    f.write(s)
```

**Function to find any user with something maximum**


```python
def max_with_x(chunkies,x):
    
    max=0
    for chunk in chunkies:
        if(max<chunk[x]):
            max=chunk[x]
            chunk_with_max_x=chunk
    return chunk_with_max_x
```

**finding user with maximum posts**

```python
chunk_with_max_posts=max_with_x(chunkies,"posts")
print(chunk_with_max_posts)
```

**finding user with maximum followers**

```python
chunk_with_max_followers=max_with_x(chunkies,"followers")
print(chunk_with_max_followers)
```

**finding user with maximum following**

```python
chunk_with_max_following=max_with_x(chunkies,"following")
print(chunk_with_max_following)
```
